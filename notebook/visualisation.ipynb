```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"D:\Jupiter\hospital_healthcare\data\clean_healthcare.csv")
```

```python
ax = sns.histplot(data=df,x="Gender")
ax.bar_label(
    ax.containers[0]
)
plt.show()
```

```python
ax = sns.histplot(data=df, x="Age", bins=6)
ax.bar_label(ax.containers[0])
plt.show()
```

```python
ax = sns.histplot(data=df,x="Blood Type", bins=5, kde=True)
ax.bar_label(ax.containers[0])
plt.show()
```

```python
df["age_group"] = pd.cut(
    df["Age"],bins=[0,18,30,45,60,100],
    labels=["0-18","19-30","31-45","46-60","61-100"])

ax = sns.countplot(
    data=df,
    x="age_group",
    hue="Gender"
)
for container_1 in ax.containers:
    ax.bar_label(container_1, fmt="%.0f",padding=6,fontsize=6)

plt.tight_layout()
    
plt.show()
```

# 🚑 Admission Analysis


```python
sns.histplot(data=df,x="Admission Type",kde=True)
```

```python
plt.figure(figsize=
           (8,6))
ax = sns.barplot(data=df,x="Admission Type",y="Billing Amount",errorbar=None)

for col in ax.containers:
    ax.bar_label(
        col,
        fmt="%.2f",
        padding=5
    )
```

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
df["stay_Days"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days
```

```python

ax = sns.barplot(
    data=df,
    x="Admission Type",
    y="stay_Days",
    estimator="mean",
    errorbar=None
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.2f",
        padding=5
    )
```

```python
ax = sns.countplot(
    data=df,
    x="Medical Condition",
    hue="Admission Type"
)

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.0f",
        padding=3,
        fontsize=7
    )
```
