```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"D:\Jupiter\hospital_healthcare\data\clean_healthcare.csv")
```

```python
ax = sns.histplot(data=df,x="Gender")
ax.bar_label(
    ax.containers[0]
)
plt.show()
```

```python
ax = sns.histplot(data=df, x="Age", bins=6)
ax.bar_label(ax.containers[0])
plt.show()
```

```python
ax = sns.histplot(data=df,x="Blood Type", bins=5, kde=True)
ax.bar_label(ax.containers[0])
plt.show()
```

```python
df["age_group"] = pd.cut(
    df["Age"],bins=[0,18,30,45,60,100],
    labels=["0-18","19-30","31-45","46-60","61-100"])

ax = sns.countplot(
    data=df,
    x="age_group",
    hue="Gender"
)
for container_1 in ax.containers:
    ax.bar_label(container_1, fmt="%.0f",padding=6,fontsize=6)

plt.tight_layout()
    
plt.show()
```

# 🩺 Medical Conditions


```python
plt.figure(figsize=(8,6))
sns.histplot(data=df,x="Medical Condition",kde=True)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(8,6))
ax = sns.countplot(data=df,x="Medical Condition",hue="Gender")
plt.xticks(rotation=90)
for container_2 in ax.containers:
    ax.bar_label(container_2,fmt="%.0f",padding=6,fontsize=6)
    
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
ax = sns.countplot(data=df,x="age_group", hue="Medical Condition")
for container_3 in ax.containers:
    ax.bar_label(container_3,fmt="%.0f",padding=8,fontsize=6)

plt.show()
```
