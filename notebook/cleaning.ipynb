# Healthcare Patient Analytics


```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"D:\Jupiter\hospital_healthcare\data\messy.csv", encoding="UTF-8")
```

```python
print(df.shape)
print(df.columns)
df = df.drop_duplicates().reset_index(drop=True)
```

```python
print(df.shape)
print(df.duplicated().sum())
df.drop_duplicates(inplace=True)
print(df.duplicated().sum())
```

```python
print(df["Name"].unique())
print(df["Name"].nunique())
print(df["Name"].value_counts())
```

*Name*

```python
df["Name"] = df["Name"].str.strip().str.lower().str.title()
df["Name"] = df["Name"].replace(["Unknown", "?", "", "NaN"], np.nan)
print(df.shape)
df.dropna(subset=["Name"], inplace=True)
df.reset_index(drop=True, inplace=True)
print(df.shape)

```

*Age*

```python
print(df["Age"].value_counts())
print(df["Age"].unique())
df["Age"] = df["Age"].astype(str).str.strip().str.lower()
df["Age"] = df["Age"].str.replace(r"\s*(years?|yrs?)",
                                  "",
                                  regex=True)
df["Age"] = df["Age"].replace(["nan", "unknown", "null","n/a","?", ""], np.nan)
df["Age"] = df["Age"].replace("twenty five",25)
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df.loc[(df["Age"] < 1) | (df["Age"] > 90), "Age"] = np.nan
print(df["Age"].isnull().sum())
df["Age"] = df["Age"].fillna(df["Age"].median()).astype(int)
print(df["Age"].unique())
```

```python
print(df["Gender"].unique())
df["Gender"] = df["Gender"].str.strip().str.lower().str.title()
df["Gender"] = df["Gender"].replace(["?", "N/A", "Null","Unknown",""], np.nan)
df["Gender"] = df["Gender"].replace({
    "F": "Female",
    "Woman": "Female",
    "Femle": "Female",
    "M": "Male",
    "Man": "Male",
    "Male.": "Male"
})
df["Gender"] = df["Gender"].fillna("Unknown")
print(df["Gender"].isnull().sum())
print(df["Gender"].unique())
```

```python
print(df["Blood Type"].unique())
df["Blood Type"] = df["Blood Type"].str.strip().str.lower().str.title()
df["Blood Type"] = df["Blood Type"].replace({"POSITIVE":"+",
                                             "positive":"-",
                                             "NEGATIVE":"-",
                                             "negative":"-",
                                             "Ab Positive":"AB+",
                                             "Ab Negative":"AB-",
                                             "A Positive":"A+",
                                             "A Negative":"A-",
                                             "Ab+":"AB+",
                                             "Ab-":"AB-",
                                             "B Negative":"B-",
                                             "B Positive":"B+",
                                             "0+":"O+",
                                             "A -":"A-",
                                             "Ab -":"AB-",
                                             "Ab +":"AB+",
                                             "B +":"B+",
                                             "B -":"B-",
                                             "O -":"O-",
                                             "O Positive":"O+",
                                             "O Negative":"O-",
                                             "O +":"O+",
                                             "A B+":"AB+",
                                             "0-":"O-",
                                             "A +":"A+"})
df["Blood Type"] = df["Blood Type"].replace(["?","Unknown","N/A", "Null", "","A","O"], np.nan)
print(df["Blood Type"].isnull().sum())
df["Blood Type"] = df["Blood Type"].fillna("Unknown")
print(df["Blood Type"].isnull().sum())
print(df["Blood Type"].unique())
```

```python
print(df["Medical Condition"].unique())
df["Medical Condition"] = df["Medical Condition"].str.strip().str.lower().str.title()
df["Medical Condition"] = df["Medical Condition"].replace(["?","None","Null","N/A", "", np.nan],"Unknown")
df["Medical Condition"] = df["Medical Condition"].replace({"Hypertention":"Hypertension",
                                                           "Canser":"Cancer",
                                                           "Astma":"Asthma",
                                                           "Diabtes":"Diabetes",
                                                           "Obsety":"Obesity",
                                                           "Arthrits":"Arthritis",
                                                           })
print(df["Medical Condition"].unique())
print(df["Medical Condition"].isnull().sum())
```

*Date of Admission*

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"] , errors="coerce", dayfirst=True, format="mixed")
print(df["Date of Admission"].value_counts())
print(df["Date of Admission"].isnull().sum())
```

```python
print(df["Doctor"].value_counts())
print(df["Doctor"].isnull().sum())
df["Doctor"] = df["Doctor"].str.strip().str.lower().str.title()
df["Doctor"] = df["Doctor"].replace({"Dr ?":"Unknown",
                                    np.nan:"Unknown"})

mask = df["Doctor"] != "Unknown"
df.loc[mask, "Doctor"] = "Dr. " + df.loc[mask, "Doctor"]
print(df["Doctor"].value_counts())
print(df["Doctor"].isnull().sum())
```

*Hospital*

```python
print(df["Hospital"].unique())
print(df["Hospital"].isnull().sum())
print(df["Hospital"].value_counts())
df["Hospital"] = df["Hospital"].str.strip().str.lower().str.title()
df["Hospital"] = df["Hospital"].replace(["?",np.nan],"Unknown Hospital")

print(df["Hospital"].unique())
print(df["Hospital"].isnull().sum())
print(df["Hospital"].value_counts())
```

```python
print(df["Insurance Provider"].unique())
df["Insurance Provider"] = df["Insurance Provider"].astype("string").str.strip().str.lower().str.title()
df["Insurance Provider"] = df["Insurance Provider"].replace(["?","N/A","Null","<NA>",""],"Unknown")
df["Insurance Provider"] = df["Insurance Provider"].replace({"Medicare.":"Medicare",
                                                             "Cigna.":"Cigna",
                                                             "Blue Cross.":"Blue Cross",
                                                             "Blue Cross Inc.":"Blue Cross",
                                                             "Bluecross":"Blue Cross",
                                                             "Cigna Inc.":"Cigna",
                                                             "Medicare Inc.":"Medicare",
                                                             "Aetna Inc":"Aetna",
                                                             "Unitedhealthcare.":"Unitedhealthcare",
                                                             "Unitedhealthcare Inc.":"Unitedhealthcare",
                                                             "Aetna.":"Aetna",
                                                             "Aetna Inc.":"Aetna",
                                                             "Medicare Inc":"Medicare"
                                                             })
df["Insurance Provider"] = df["Insurance Provider"].fillna("Unknown")
print(df["Insurance Provider"].isna().sum())
```

```python
df["Billing Amount"] = df["Billing Amount"].astype("string").str.strip()

df["Billing Amount"] = df["Billing Amount"].replace(
    ["?", "N/A", "Null", "unknown", "Unknown", ""],
    pd.NA
)
df["Billing Amount"] = (
    df["Billing Amount"]
    .str.replace(",", "", regex=False)
    .str.replace("₹", "", regex=False)
    .str.replace("INR", "", regex=False)
    .str.replace("USD", "", regex=False)
)
df["Billing Amount"] = df["Billing Amount"].str.replace(
    r"^(\d+(?:\.\d+)?)k$",
    lambda x: str(float(x.group(1)) * 1000),
    regex=True
)
print(df["Billing Amount"].isnull().sum())

df["Billing Amount"] = pd.to_numeric(
    df["Billing Amount"],
    errors="coerce"
).round(2)
print(df.loc[df["Billing Amount"] < 0,"Billing Amount"], np.nan)
df["Billing Amount"] = df["Billing Amount"].fillna(df["Billing Amount"].mean())
print(df["Billing Amount"].isnull().sum())
print(df["Billing Amount"].unique())
```

*Room Number*

```python
print(df["Room Number"].unique())
print(df["Room Number"].value_counts())
df["Room Number"] = df["Room Number"].astype("string").str.strip().str.lower().str.title()
df["Room Number"] = (df["Room Number"].str.replace("A","",regex=False)
                     .str.replace("#","",regex=False)
                     .str.replace("R-","",regex=False)
                     .str.replace("Room ","",regex=False)
)
df["Room Number"] = df["Room Number"].replace(["?","Null","N/",""],np.nan)
df["Room Number"] = pd.to_numeric(df["Room Number"],errors="coerce")
df.loc[~df["Room Number"].between(101,999),"Room Number"]  = np.nan
print(df["Room Number"].unique())
print(df["Room Number"].value_counts())
```

```python
print(df["Admission Type"].unique())
df["Admission Type"] = df["Admission Type"].str.strip().str.lower().str.title()
df["Admission Type"] = df["Admission Type"].replace(["Null","N/A","?","Unknown",""],np.nan)
df["Admission Type"] = df["Admission Type"].replace({"Emerg.":"Emergency",
                                                     "Emerg":"Emergency",
                                                     "Er":"Emergency",
                                                     "Urgent":"Emergency",
                                                     "Planned":"Scheduled",
                                                     "Elective":"Scheduled"
                                                     })
df["Admission Type"] = df["Admission Type"].fillna("Unknown")
```

```python
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"],errors="coerce",dayfirst=True)

print(df["Discharge Date"].unique())
print(df["Discharge Date"].value_counts())
```

```python
print(df["Medication"].unique())
df["Medication"] = df["Medication"].astype("string").str.strip().str.lower().str.title()
df["Medication"] = df["Medication"].replace(["?","Unknown","Null","","None","N/A"],pd.NA)
df["Medication"] = df["Medication"].replace({"Paracetamo":"Paracetamol",
                                             "Lipito":"Lipitor",
                                             "Penicilli":"Penicillin",
                                             "Ibuprofe":"Ibuprofen",
                                             "Aspiri":"Aspirin",
                                             })
print(df["Medication"].isna().sum())
```

```python
print(df["Test Results"].unique())
df["Test Results"] = df["Test Results"].astype("string").str.strip().str.lower().str.title()
df["Test Results"] = df["Test Results"].replace(["?","Null","","N/A","Unknown","Not Available"],pd.NA)
df["Test Results"] = df["Test Results"].replace({"Abn":"Abnormal",
                                                 "Norm":"Normal"})
```

```python
df.to_csv("D:\Jupiter\hospital_healthcare\data\clean_healthcare.csv", index=False)
```
