```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"D:\Jupiter\hospital_healthcare\data\messy.csv", encoding="UTF-8")
```

```python
print(df.shape)
print(df.columns)
```

```python
print(df.shape)
print(df.duplicated().sum())
df.drop_duplicates(inplace=True)
print(df.duplicated().sum())
```

```python
print(df["Name"].unique())
print(df["Name"].nunique())
print(df["Name"].value_counts())
```

*Name*

```python
df["Name"] = df["Name"].str.strip().str.lower().str.title()
df["Name"] = df["Name"].replace(["Unknown", "?", "", "NaN"], np.nan)
print(df.shape)
df.dropna(subset=["Name"], inplace=True)
df.reset_index(drop=True, inplace=True)
print(df.shape)

```

*Age*

```python
# print(df["Age"].value_counts())
# print(df["Age"].unique())
df["Age"] = df["Age"].astype(str).str.strip().str.lower()
df["Age"] = df["Age"].str.replace(r"\s*(years?|yrs?)",
                                  "",
                                  regex=True)
df["Age"] = df["Age"].replace(["nan", "unknown", "null","n/a","?", ""], np.nan)
df["Age"] = df["Age"].replace("twenty five",25)
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df.loc[(df["Age"] < 1) | (df["Age"] > 90), "Age"] = np.nan
print(df["Age"].isnull().sum())
df["Age"] = df["Age"].fillna(df["Age"].median()).astype(int)
print(df["Age"].unique())
```

```python
print(df["Gender"].unique())
df["Gender"] = df["Gender"].str.strip().str.lower().str.title()
df["Gender"] = df["Gender"].replace(["?", "N/A", "Null","Unknown",""], np.nan)
df["Gender"] = df["Gender"].replace({
    "F": "Female",
    "Woman": "Female",
    "Femle": "Female",
    "M": "Male",
    "Man": "Male",
    "Male.": "Male"
})
df["Gender"] = df["Gender"].fillna("Unknown")
print(df["Gender"].isnull().sum())
print(df["Gender"].unique())
```

```python
# print(df["Blood Type"].unique())
df["Blood Type"] = df["Blood Type"].str.strip().str.lower().str.title()
df["Blood Type"] = df["Blood Type"].replace({"POSITIVE":"+",
                                             "positive":"-",
                                             "NEGATIVE":"-",
                                             "negative":"-",
                                             "Ab Positive":"AB+",
                                             "Ab Negative":"AB-",
                                             "A Positive":"A+",
                                             "A Negative":"A-",
                                             "Ab+":"AB+",
                                             "Ab-":"AB-",
                                             "B Negative":"B-",
                                             "B Positive":"B+",
                                             "0+":"O+",
                                             "A -":"A-",
                                             "Ab -":"AB-",
                                             "Ab +":"AB+",
                                             "B +":"B+",
                                             "B -":"B-",
                                             "O -":"O-",
                                             "O Positive":"O+",
                                             "O Negative":"O-",
                                             "O +":"O+",
                                             "A B+":"AB+",
                                             "0-":"O-",
                                             "A +":"A+"})
df["Blood Type"] = df["Blood Type"].replace(["?","Unknown","N/A", "Null", "","A","O"], np.nan)
print(df["Blood Type"].isnull().sum())
df["Blood Type"] = df["Blood Type"].fillna("Unknown")
print(df["Blood Type"].isnull().sum())
print(df["Blood Type"].unique())
```

```python
# print(df["Medical Condition"].unique())
df["Medical Condition"] = df["Medical Condition"].str.strip().str.lower().str.title()
df["Medical Condition"] = df["Medical Condition"].replace(["?","None","Null","N/A", "", np.nan],"Unknown")
df["Medical Condition"] = df["Medical Condition"].replace({"Hypertention":"Hypertension",
                                                           "Canser":"Cancer",
                                                           "Astma":"Asthma",
                                                           "Diabtes":"Diabetes",
                                                           "Obsety":"Obesity",
                                                           "Arthrits":"Arthritis",
                                                           })
print(df["Medical Condition"].unique())
print(df["Medical Condition"].isnull().sum())
```

*Date of Admission*

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"] , errors="coerce", dayfirst=True, format="mixed")
print(df["Date of Admission"].value_counts())
print(df["Date of Admission"].isnull().sum())
```

```python
# print(df["Doctor"].value_counts())
# print(df["Doctor"].isnull().sum())
df["Doctor"] = df["Doctor"].str.strip().str.lower().str.title()
df["Doctor"] = df["Doctor"].replace({"Dr ?":"Unknown",
                                    np.nan:"Unknown"})

mask = df["Doctor"] != "Unknown"
df.loc[mask, "Doctor"] = "Dr. " + df.loc[mask, "Doctor"]
print(df["Doctor"].value_counts())
print(df["Doctor"].isnull().sum())
```

*Hospital*

```python
# print(df["Hospital"].unique())
print(df["Hospital"].isnull().sum())
print(df["Hospital"].value_counts())
df["Hospital"] = df["Hospital"].replace(["?",np.nan],"Unknown Hospital")
```

```python
print(df["Insurance Provider"].unique())
df["Insurance Provider"] = df["Insurance Provider"].astype("string").str.strip().str.lower().str.title()
df["Insurance Provider"] = df["Insurance Provider"].replace(["?","N/A","Null","<NA>",""],"Unknown")
df["Insurance Provider"] = df["Insurance Provider"].replace({"Medicare.":"Medicare",
                                                             "Cigna.":"Cigna",
                                                             "Blue Cross.":"Blue Cross",
                                                             "Blue Cross Inc.":"Blue Cross",
                                                             "Bluecross":"Blue Cross",
                                                             "Cigna Inc.":"Cigna",
                                                             "Medicare Inc.":"Medicare",
                                                             "Aetna Inc":"Aetna",
                                                             "Unitedhealthcare.":"Unitedhealthcare",
                                                             "Unitedhealthcare Inc.":"Unitedhealthcare",
                                                             "Aetna.":"Aetna",
                                                             "Aetna Inc.":"Aetna",
                                                             "Medicare Inc":"Medicare"
                                                             })
df["Insurance Provider"] = df["Insurance Provider"].fillna("Unknown")
print(df["Insurance Provider"].isna().sum())
```

```python
df["Billing Amount"] = df["Billing Amount"].astype("string").str.strip()

df["Billing Amount"] = df["Billing Amount"].replace(
    ["?", "N/A", "Null", "unknown", "Unknown", ""],
    pd.NA
)
df["Billing Amount"] = (
    df["Billing Amount"]
    .str.replace(",", "", regex=False)
    .str.replace("₹", "", regex=False)
    .str.replace("INR", "", regex=False)
    .str.replace("USD", "", regex=False)
)
df["Billing Amount"] = df["Billing Amount"].str.replace(
    r"^(\d+(?:\.\d+)?)k$",
    lambda x: str(float(x.group(1)) * 1000),
    regex=True
)
df["Billing Amount"] = pd.to_numeric(
    df["Billing Amount"],
    errors="coerce"
).round(2)
print(df["Billing Amount"].unique())
```

*Room Number*

```python
# print(df["Room Number"].unique())
# print(df["Room Number"].value_counts())
df["Room Number"] = df["Room Number"].astype("string").str.strip().str.lower().str.title()
df["Room Number"] = (df["Room Number"].str.replace("A","",regex=False)
                     .str.replace("#","",regex=False)
                     .str.replace("R-","",regex=False)
                     .str.replace("Room ","",regex=False)
)
df["Room Number"] = df["Room Number"].replace(["?","Null","N/",""],np.nan)
df["Room Number"] = pd.to_numeric(df["Room Number"],errors="coerce")
df.loc[~df["Room Number"].between(101,999),"Room Number"]  = np.nan
# print(df["Room Number"].unique())
# print(df["Room Number"].value_counts())
```

```python
print(df["Admission Type"].unique())
df["Admission Type"] = df["Admission Type"].str.strip().str.lower().str.title()
df["Admission Type"] = df["Admission Type"].replace(["Null","N/A","?","Unknown",""],np.nan)
df["Admission Type"] = df["Admission Type"].replace({"Emerg.":"Emergency",
                                                     "Emerg":"Emergency",
                                                     "Er":"Emergency",
                                                     "Urgent":"Emergency",
                                                     "Planned":"Scheduled",
                                                     "Elective":"Scheduled"
                                                     })
df["Admission Type"] = df["Admission Type"].fillna("Unknown")
```

```python
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"],errors="coerce",dayfirst=True)

print(df["Discharge Date"].unique())
print(df["Discharge Date"].value_counts())
```

```python
print(df["Medication"].unique())
df["Medication"] = df["Medication"].astype("string").str.strip().str.lower().str.title()
df["Medication"] = df["Medication"].replace(["?","Unknown","Null","","None","N/A"],pd.NA)
df["Medication"] = df["Medication"].replace({"Paracetamo":"Paracetamol",
                                             "Lipito":"Lipitor",
                                             "Penicilli":"Penicillin",
                                             "Ibuprofe":"Ibuprofen",
                                             "Aspiri":"Aspirin",
                                             })
# print(df["Medication"].isna().sum())
```

```python
print(df["Test Results"].unique())
df["Test Results"] = df["Test Results"].astype("string").str.strip().str.lower().str.title()
df["Test Results"] = df["Test Results"].replace(["?","Null","","N/A","Unknown","Not Available"],pd.NA)
df["Test Results"] = df["Test Results"].replace({"Abn":"Abnormal",
                                                 "Norm":"Normal"})
```

```python
import pandas as pd
import numpy as np

# ============================================================
#        HEALTHCARE DATASET - COMPLETE CLEANING CHECK
# ============================================================

# Your columns:
# Name, Age, Gender, Blood Type, Medical Condition,
# Date of Admission, Doctor, Hospital, Insurance Provider,
# Billing Amount, Room Number, Admission Type,
# Discharge Date, Medication, Test Results


# ------------------------------------------------------------
# 1. CLEAN COLUMN NAMES
# ------------------------------------------------------------

df.columns = (
    df.columns
    .str.strip()
    .str.replace("*", "", regex=False)
)

print("=" * 80)
print("           HEALTHCARE DATASET - CLEANING CHECK")
print("=" * 80)


# ------------------------------------------------------------
# 2. BASIC INFORMATION
# ------------------------------------------------------------

print("\n📊 BASIC INFORMATION")
print("-" * 80)

print("Rows       :", len(df))
print("Columns    :", len(df.columns))
print("Duplicates :", df.duplicated().sum())


# ------------------------------------------------------------
# 3. EXPECTED COLUMNS
# ------------------------------------------------------------

expected_columns = [
    "Name",
    "Age",
    "Gender",
    "Blood Type",
    "Medical Condition",
    "Date of Admission",
    "Doctor",
    "Hospital",
    "Insurance Provider",
    "Billing Amount",
    "Room Number",
    "Admission Type",
    "Discharge Date",
    "Medication",
    "Test Results"
]

print("\n📋 COLUMN CHECK")
print("-" * 80)

for col in expected_columns:

    if col in df.columns:
        print(f"✅ {col}")
    else:
        print(f"❌ MISSING COLUMN: {col}")


# ------------------------------------------------------------
# 4. GENERAL COLUMN REPORT
# ------------------------------------------------------------

print("\n📋 COLUMN-WISE REPORT")
print("-" * 80)

for col in expected_columns:

    if col not in df.columns:
        continue

    missing = df[col].isna().sum()
    missing_pct = df[col].isna().mean() * 100
    unique = df[col].nunique(dropna=True)

    print(
        f"{col:<22} "
        f"Missing: {missing:<5} "
        f"({missing_pct:>6.2f}%) "
        f"Unique: {unique}"
    )


# ============================================================
# 5. COLUMN-SPECIFIC VALIDATION
# ============================================================

print("\n" + "=" * 80)
print("              COLUMN-SPECIFIC VALIDATION")
print("=" * 80)


# ------------------------------------------------------------
# NAME
# ------------------------------------------------------------

print("\n👤 Name")

invalid_name = (
    df["Name"].notna() &
    (
        df["Name"].astype("string").str.strip().eq("") |
        df["Name"].astype("string").isin(
            ["?", "Null", "N/A", "Unknown", "None"]
        )
    )
).sum()

print(
    "Status:",
    "✅ Clean" if invalid_name == 0
    else f"❌ {invalid_name} invalid values"
)


# ------------------------------------------------------------
# AGE
# ------------------------------------------------------------

print("\n🎂 Age")

# Check numeric
age_numeric = pd.to_numeric(
    df["Age"],
    errors="coerce"
)

invalid_age = (
    df["Age"].notna() &
    age_numeric.isna()
).sum()

out_of_range_age = (
    age_numeric.notna() &
    ~age_numeric.between(0, 120)
).sum()

print(
    "Numeric:",
    "✅" if invalid_age == 0
    else f"❌ {invalid_age} non-numeric"
)

print(
    "Range 0-120:",
    "✅" if out_of_range_age == 0
    else f"❌ {out_of_range_age} invalid"
)


# ------------------------------------------------------------
# GENDER
# ------------------------------------------------------------

print("\n🚻 Gender")

valid_gender = [
    "Male",
    "Female",
    "Other"
]

invalid_gender = (
    df["Gender"].notna() &
    ~df["Gender"].isin(valid_gender)
).sum()

print(
    "Status:",
    "✅ Clean" if invalid_gender == 0
    else f"❌ {invalid_gender} invalid values"
)

print("Values:")
print(df["Gender"].value_counts(dropna=False))


# ------------------------------------------------------------
# BLOOD TYPE
# ------------------------------------------------------------

print("\n🩸 Blood Type")

valid_blood = [
    "A+", "A-",
    "B+", "B-",
    "AB+", "AB-",
    "O+", "O-"
]

invalid_blood = (
    df["Blood Type"].notna() &
    ~df["Blood Type"].isin(valid_blood)
).sum()

print(
    "Status:",
    "✅ Clean" if invalid_blood == 0
    else f"❌ {invalid_blood} invalid values"
)

print("Values:")
print(df["Blood Type"].value_counts(dropna=False))


# ------------------------------------------------------------
# MEDICAL CONDITION
# ------------------------------------------------------------

print("\n🏥 Medical Condition")

invalid_medical = (
    df["Medical Condition"]
    .astype("string")
    .str.strip()
    .isin(["?", "Null", "N/A", "Unknown", ""])
).sum()

print(
    "Status:",
    "✅ Clean" if invalid_medical == 0
    else f"❌ {invalid_medical} invalid values"
)

print("Values:")
print(df["Medical Condition"].value_counts(dropna=False))


# ------------------------------------------------------------
# DATE OF ADMISSION
# ------------------------------------------------------------

print("\n📅 Date of Admission")

admission_date = pd.to_datetime(
    df["Date of Admission"],
    errors="coerce",
    format="mixed",
    dayfirst=True
)

invalid_admission_date = (
    df["Date of Admission"].notna() &
    admission_date.isna()
).sum()

print(
    "Datetime:",
    "✅ Clean" if invalid_admission_date == 0
    else f"❌ {invalid_admission_date} invalid dates"
)

print("Min:", admission_date.min())
print("Max:", admission_date.max())


# ------------------------------------------------------------
# DOCTOR
# ------------------------------------------------------------

print("\n👨‍⚕️ Doctor")

invalid_doctor = (
    df["Doctor"]
    .astype("string")
    .str.strip()
    .isin(["?", "Null", "N/A", "Unknown", ""])
).sum()

print(
    "Status:",
    "✅ Clean" if invalid_doctor == 0
    else f"❌ {invalid_doctor} invalid values"
)


# ------------------------------------------------------------
# HOSPITAL
# ------------------------------------------------------------

print("\n🏥 Hospital")

invalid_hospital = (
    df["Hospital"]
    .astype("string")
    .str.strip()
    .isin(["?", "Null", "N/A", "Unknown", ""])
).sum()

print(
    "Status:",
    "✅ Clean" if invalid_hospital == 0
    else f"❌ {invalid_hospital} invalid values"
)


# ------------------------------------------------------------
# INSURANCE PROVIDER
# ------------------------------------------------------------

print("\n💳 Insurance Provider")

invalid_insurance = (
    df["Insurance Provider"]
    .astype("string")
    .str.strip()
    .isin(["?", "Null", "N/A", "Unknown", ""])
).sum()

print(
    "Status:",
    "✅ Clean" if invalid_insurance == 0
    else f"❌ {invalid_insurance} invalid values"
)


# ------------------------------------------------------------
# BILLING AMOUNT
# ------------------------------------------------------------

print("\n💰 Billing Amount")

billing_numeric = pd.to_numeric(
    df["Billing Amount"],
    errors="coerce"
)

invalid_billing_type = (
    df["Billing Amount"].notna() &
    billing_numeric.isna()
).sum()

negative_billing = (
    billing_numeric.notna() &
    (billing_numeric < 0)
).sum()

print(
    "Numeric:",
    "✅" if invalid_billing_type == 0
    else f"❌ {invalid_billing_type} non-numeric"
)

print(
    "Negative:",
    "✅" if negative_billing == 0
    else f"❌ {negative_billing} negative values"
)


# ------------------------------------------------------------
# ROOM NUMBER
# ------------------------------------------------------------

print("\n🚪 Room Number")

room_numeric = pd.to_numeric(
    df["Room Number"],
    errors="coerce"
)

invalid_room_type = (
    df["Room Number"].notna() &
    room_numeric.isna()
).sum()

invalid_room_range = (
    room_numeric.notna() &
    ~room_numeric.between(101, 999)
).sum()

print(
    "Numeric:",
    "✅" if invalid_room_type == 0
    else f"❌ {invalid_room_type} non-numeric"
)

print(
    "Range 101-999:",
    "✅" if invalid_room_range == 0
    else f"❌ {invalid_room_range} invalid"
)


# ------------------------------------------------------------
# ADMISSION TYPE
# ------------------------------------------------------------

print("\n🚨 Admission Type")

valid_admission = [
    "Emergency",
    "Scheduled",
    "Priority"
]

invalid_admission = (
    df["Admission Type"].notna() &
    ~df["Admission Type"].isin(valid_admission)
).sum()

print(
    "Status:",
    "✅ Clean" if invalid_admission == 0
    else f"❌ {invalid_admission} invalid values"
)

print("Values:")
print(df["Admission Type"].value_counts(dropna=False))


# ------------------------------------------------------------
# DISCHARGE DATE
# ------------------------------------------------------------

print("\n📅 Discharge Date")

discharge_date = pd.to_datetime(
    df["Discharge Date"],
    errors="coerce",
    format="mixed",
    dayfirst=True
)

invalid_discharge_date = (
    df["Discharge Date"].notna() &
    discharge_date.isna()
).sum()

print(
    "Datetime:",
    "✅ Clean" if invalid_discharge_date == 0
    else f"❌ {invalid_discharge_date} invalid dates"
)

# Compare admission and discharge
date_order_error = (
    admission_date.notna() &
    discharge_date.notna() &
    (discharge_date < admission_date)
).sum()

print(
    "Date order:",
    "✅ Correct" if date_order_error == 0
    else f"❌ {date_order_error} discharge dates before admission"
)


# ------------------------------------------------------------
# MEDICATION
# ------------------------------------------------------------

print("\n💊 Medication")

invalid_medication = (
    df["Medication"]
    .astype("string")
    .str.strip()
    .isin(["?", "Unknown", "Null", "None", "N/A", ""])
).sum()

print(
    "Status:",
    "✅ Clean" if invalid_medication == 0
    else f"❌ {invalid_medication} invalid values"
)

print("Values:")
print(df["Medication"].value_counts(dropna=False))


# ------------------------------------------------------------
# TEST RESULTS
# ------------------------------------------------------------

print("\n🧪 Test Results")

valid_results = [
    "Normal",
    "Abnormal",
    "Positive",
    "Negative",
    "Pending",
    "Inconclusive"
]

invalid_results = (
    df["Test Results"].notna() &
    ~df["Test Results"].isin(valid_results)
).sum()

print(
    "Status:",
    "✅ Clean" if invalid_results == 0
    else f"❌ {invalid_results} invalid values"
)

print("Values:")
print(df["Test Results"].value_counts(dropna=False))


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("                    🎯 FINAL SUMMARY")
print("=" * 80)

checks = {
    "Name": invalid_name,
    "Age Type": invalid_age,
    "Age Range": out_of_range_age,
    "Gender": invalid_gender,
    "Blood Type": invalid_blood,
    "Medical Condition": invalid_medical,
    "Admission Date": invalid_admission_date,
    "Doctor": invalid_doctor,
    "Hospital": invalid_hospital,
    "Insurance Provider": invalid_insurance,
    "Billing Type": invalid_billing_type,
    "Negative Billing": negative_billing,
    "Room Type": invalid_room_type,
    "Room Range": invalid_room_range,
    "Admission Type": invalid_admission,
    "Discharge Date": invalid_discharge_date,
    "Date Order": date_order_error,
    "Medication": invalid_medication,
    "Test Results": invalid_results
}

for check, count in checks.items():

    if count == 0:
        print(f"✅ {check:<25} Clean")
    else:
        print(f"❌ {check:<25} {count} issue(s)")

print("=" * 80)

total_issues = sum(checks.values())

if total_issues == 0:
    print("🎉 ALL COLUMN VALIDATION CHECKS PASSED!")
else:
    print(f"⚠️ TOTAL ISSUES FOUND: {total_issues}")

print("=" * 80)
```
