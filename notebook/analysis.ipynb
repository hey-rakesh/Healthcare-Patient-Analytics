# Level 1: Basic Patient Analysis

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"D:\Jupiter\hospital_healthcare\data\clean_healthcare.csv")
print(df.head())
```

```python
print(df["Name"].size)
print(df["Gender"].value_counts())
print(df["Age"].mean())
print(df["Age"].min())
print(df["Age"].max())
print(df["Blood Type"].value_counts())
print(df["Hospital"].unique())
print(df["Doctor"].unique())
print(df["Insurance Provider"].unique())
print(df["Blood Type"].value_counts().idxmax())
```

# Level 2: Medical Condition Analysis 🩺

```python
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
patient_medi_cond = df.groupby("Medical Condition")["Name"].count()
avg_age_medi_cond = df.groupby("Medical Condition")["Age"].mean().round()
avg_bill_medi_cond = df.groupby("Medical Condition")["Billing Amount"].mean()

df["Date of Admission"] = pd.to_datetime(df["Date of Admission"],errors="coerce")
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"], errors="coerce")
df["stay_days"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days
avg_day_medi_cond = df.groupby("Medical Condition")["stay_days"].mean().round()

high_avg_bill_medi_cond = df.groupby("Medical Condition")["Billing Amount"].mean().idxmax()
high_avg_day_medi_cond = df.groupby("Medical Condition")["stay_days"].mean().round()

gender_medi_cond = df.groupby("Medical Condition")["Gender"].value_counts()
gender_medi = pd.crosstab(df["Medical Condition"],df["Gender"])

df["Age Group"] = pd.cut(df["Age"], bins=[0, 18, 30, 45, 60, 100],labels=["Child", "Young", "Adult", "Middle Age", "Senior"])
age_cond = pd.crosstab(df["Age Group"], df["Medical Condition"])
most_common = age_cond.idxmax(axis=1)
```
