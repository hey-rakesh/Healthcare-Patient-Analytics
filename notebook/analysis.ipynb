# Level 1: Basic Patient Analysis

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"D:\Jupiter\hospital_healthcare\data\clean_healthcare.csv")
print(df.head())
```

```python
print(df["Name"].size)
print(df["Gender"].value_counts())
print(df["Age"].mean())
print(df["Age"].min())
print(df["Age"].max())
print(df["Blood Type"].value_counts())
print(df["Hospital"].unique())
print(df["Doctor"].unique())
print(df["Insurance Provider"].unique())
print(df["Blood Type"].value_counts().idxmax())
```

# Level 2: Medical Condition Analysis 🩺

```python
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
patient_medi_cond = df.groupby("Medical Condition")["Name"].count()
avg_age_medi_cond = df.groupby("Medical Condition")["Age"].mean().round()
avg_bill_medi_cond = df.groupby("Medical Condition")["Billing Amount"].mean()

df["Date of Admission"] = pd.to_datetime(df["Date of Admission"],errors="coerce")
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"], errors="coerce")
df["stay_days"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days
avg_day_medi_cond = df.groupby("Medical Condition")["stay_days"].mean().round()

high_avg_bill_medi_cond = df.groupby("Medical Condition")["Billing Amount"].mean().idxmax()
high_avg_day_medi_cond = df.groupby("Medical Condition")["stay_days"].mean().round()

gender_medi_cond = df.groupby("Medical Condition")["Gender"].value_counts()
gender_medi = pd.crosstab(df["Medical Condition"],df["Gender"])

df["Age Group"] = pd.cut(df["Age"], bins=[0, 18, 30, 45, 60, 100],labels=["Child", "Young", "Adult", "Middle Age", "Senior"])
age_cond = pd.crosstab(df["Age Group"], df["Medical Condition"])
most_common = age_cond.idxmax(axis=1)
```

#  Level 3: Admission Analysis 🚑

```python
print(df["Admission Type"].value_counts())
print(df["Admission Type"].value_counts().idxmax())
print(df.groupby("Admission Type")["Billing Amount"].mean())
print(df.groupby("Admission Type")["stay_days"].mean().round().astype(int))
print(df.groupby("Admission Type")["Billing Amount"].mean().idxmax())
print(df.groupby(["Medical Condition", "Admission Type"]).size())
print(df[df["Admission Type"] == "Emergency"]["Medical Condition"].value_counts().head(1))
```

# Level 4: Hospital Analysis 🏥


```python
hospital_patients = df.groupby("Hospital").size()
top_10_hospitals = df["Hospital"].value_counts().head(10)
print(df.groupby("Hospital")["Billing Amount"].mean().idxmax())
print(df.groupby("Hospital")["stay_days"].mean().idxmax())
print(df.groupby("Hospital")["Billing Amount"].sum().sort_values(ascending=False))
top_10_bill = df.groupby("Hospital")["Billing Amount"].sum().sort_values(ascending=False).head(10)
hospital_condition = pd.crosstab(df["Hospital"],df["Medical Condition"])
hospital_admission = pd.crosstab(df["Hospital"],df["Admission Type"])

```

# Level 5: Insurance Analysis 💳

```python
print(df["Insurance Provider"].unique())
print(df["Insurance Provider"].value_counts())
print(df.groupby("Insurance Provider")["Billing Amount"].mean().sort_values(ascending=False).head(1))
print(df.groupby("Insurance Provider")["Billing Amount"].mean().idxmax())
print(df.groupby("Insurance Provider")["Billing Amount"].sum())
print(pd.crosstab(df["Insurance Provider"],df["Medical Condition"]))
print(pd.crosstab(df["Insurance Provider"],df["Admission Type"]))
```
